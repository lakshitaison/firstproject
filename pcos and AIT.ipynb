```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
kegg = kegg[kegg["Adjusted P-value"] < 0.01]

print(kegg.shape)
```

```
import pandas as pd

# Collect all genes from KEGG pathways
all_genes = set()

for genes in kegg["Genes"]:
    for g in genes.split(";"):
        all_genes.add(g.strip())

all_genes = sorted(all_genes)

print("Genes in KEGG:", len(all_genes))
```

```
matrix = pd.DataFrame(
    0,
    index=kegg["Term"],
    columns=all_genes
)
```

```
for _, row in kegg.iterrows():

    pathway = row["Term"]

    genes = [g.strip() for g in row["Genes"].split(";")]

    for gene in genes:

        matrix.loc[pathway, gene] = 1
```

```
from sklearn.metrics import pairwise_distances
import pandas as pd

distance = pairwise_distances(
    matrix.values,
    metric="jaccard"
)

similarity = 1 - distance

similarity = pd.DataFrame(
    similarity,
    index=matrix.index,
    columns=matrix.index
)

similarity.head()
```

```
from scipy.cluster.hierarchy import linkage

Z = linkage(
    distance,
    method="average"
)
```

```
from scipy.cluster.hierarchy import dendrogram
import matplotlib.pyplot as plt

plt.figure(figsize=(10,8))

dendrogram(
    Z,
    labels=matrix.index,
    leaf_rotation=90,
)

plt.tight_layout()

plt.show()
```

```
import pandas as pd

enrichment = pd.read_csv("PCOS_AIT_Enrichment.csv")

kegg = enrichment[
    enrichment["Gene_set"] == "KEGG_2021_Human"
]

kegg = kegg[kegg["Adjusted P-value"] < 0.01]

print("Significant KEGG pathways:", len(kegg))

kegg.head()
```

```
# Collect all unique genes

all_genes = set()

for genes in kegg["Genes"]:
    genes = genes.split(";")
    all_genes.update([g.strip() for g in genes])

all_genes = sorted(list(all_genes))

print("Unique genes:", len(all_genes))
```

```
matrix = pd.DataFrame(
    0,
    index=kegg["Term"],
    columns=all_genes
)
```

```
for _, row in kegg.iterrows():

    pathway = row["Term"]

    genes = [g.strip() for g in row["Genes"].split(";")]

    for gene in genes:
        if gene in matrix.columns:
            matrix.loc[pathway, gene] = 1
```

```
matrix.shape
```

```
matrix.to_csv("KEGG_Pathway_Gene_Matrix.csv")
```

```
from sklearn.metrics import pairwise_distances
import pandas as pd

distance = pairwise_distances(
    matrix.values,
    metric="jaccard"
)

similarity = 1 - distance

similarity_df = pd.DataFrame(
    similarity,
    index=matrix.index,
    columns=matrix.index
)

similarity_df.head()
```

```
similarity_df.to_csv("KEGG_Jaccard_Similarity.csv")
```

```
from scipy.cluster.hierarchy import linkage

Z = linkage(
    distance,
    method="average"
)
```

```
from scipy.cluster.hierarchy import dendrogram
import matplotlib.pyplot as plt

plt.figure(figsize=(18,8))

dendrogram(
    Z,
    labels=matrix.index,
    leaf_rotation=90,
    leaf_font_size=12
)

plt.title("Hierarchical Clustering of KEGG Pathways")

plt.tight_layout()

plt.show()
```

```
# ================================
# Publication-quality Network Graph
# ================================

import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from collections import Counter

# -----------------------------
# Top hub genes
# -----------------------------
counter = Counter()

for genes in kegg["Genes"]:
    counter.update([g.strip() for g in genes.split(";")])

top_genes = [gene for gene, count in counter.most_common(20)]

# -----------------------------
# Top significant pathways
# -----------------------------
top_pathways = list(
    kegg.sort_values("Adjusted P-value")
        .head(15)["Term"]
)

# -----------------------------
# Build graph
# -----------------------------
G = nx.Graph()

for _, row in kegg.iterrows():

    pathway = row["Term"]

    if pathway not in top_pathways:
        continue

    genes = [g.strip() for g in row["Genes"].split(";")]

    for gene in genes:

        if gene in top_genes:
            G.add_edge(pathway, gene)

# -----------------------------
# Layout
# -----------------------------
pos = nx.kamada_kawai_layout(G)

# -----------------------------
# Degree-based node sizes
# -----------------------------
degree = dict(G.degree())

gene_sizes = []
pathway_sizes = []

for node in top_genes:
    if node in G.nodes():
        gene_sizes.append(300 + degree[node]*250)

for node in top_pathways:
    if node in G.nodes():
        pathway_sizes.append(1200 + degree[node]*350)

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(16,12))

# Pathway nodes
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=[n for n in top_pathways if n in G.nodes()],
    node_color="#E64B35",
    node_size=pathway_sizes,
    edgecolors="black",
    linewidths=1.5,
    alpha=0.95,
    label="KEGG Pathways"
)

# Gene nodes
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=[n for n in top_genes if n in G.nodes()],
    node_color="#4DBBD5",
    node_size=gene_sizes,
    edgecolors="black",
    linewidths=1,
    alpha=0.9,
    label="Hub Genes"
)

# Edges
nx.draw_networkx_edges(
    G,
    pos,
    edge_color="gray",
    width=2,
    alpha=0.5
)

# Labels
nx.draw_networkx_labels(
    G,
    pos,
    font_size=10,
    font_family="sans-serif"
)

plt.title(
    "Top Hub Gene–KEGG Pathway Interaction Network",
    fontsize=18,
    weight="bold"
)

plt.legend(fontsize=12)

plt.axis("off")

plt.tight_layout()

plt.savefig(
    "HubGene_KEGG_Network.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```
